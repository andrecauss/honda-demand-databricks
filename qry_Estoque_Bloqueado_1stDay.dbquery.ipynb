```sql
WITH base AS (
  SELECT
    c.item_principal_cadeia AS item_principal,
    date_format(h.reference_date, 'yyyy/MM') AS mes,
    SUM(h.estoque_bloqueado) AS estoque_bloqueado
  FROM `parts_hdbk_sandbox`.`pr_cadastrao`.`material_inventory_history` h
  INNER JOIN `parts_hdbk_sandbox`.`pr_cadastrao`.`material_cadeia` c
    ON h.material = c.material
    AND h.empresa = c.empresa
  GROUP BY 1, 2
),
pivotado AS (
  SELECT *
  FROM base
  PIVOT (
    SUM(estoque_bloqueado)
    FOR mes IN (
      '2025/06', '2025/07', '2025/08', '2025/09', '2025/10', '2025/11', '2025/12',
      '2026/01', '2026/02', '2026/03', '2026/04', '2026/05', '2026/06', '2026/07', '2026/08', '2026/09'
    )
  )
)
SELECT
  item_principal,
  COALESCE(SUM(`2025/06`), 0) AS `2025/06`,
  COALESCE(SUM(`2025/07`), 0) AS `2025/07`,
  COALESCE(SUM(`2025/08`), 0) AS `2025/08`,
  COALESCE(SUM(`2025/09`), 0) AS `2025/09`,
  COALESCE(SUM(`2025/10`), 0) AS `2025/10`,
  COALESCE(SUM(`2025/11`), 0) AS `2025/11`,
  COALESCE(SUM(`2025/12`), 0) AS `2025/12`,
  COALESCE(SUM(`2026/01`), 0) AS `2026/01`,
  COALESCE(SUM(`2026/02`), 0) AS `2026/02`,
  COALESCE(SUM(`2026/03`), 0) AS `2026/03`,
  COALESCE(SUM(`2026/04`), 0) AS `2026/04`,
  COALESCE(SUM(`2026/05`), 0) AS `2026/05`,
  COALESCE(SUM(`2026/06`), 0) AS `2026/06`,
  COALESCE(SUM(`2026/07`), 0) AS `2026/07`,
  COALESCE(SUM(`2026/08`), 0) AS `2026/08`,
  COALESCE(SUM(`2026/09`), 0) AS `2026/09`
FROM pivotado
GROUP BY item_principal
ORDER BY `2026/09` DESC;
```
